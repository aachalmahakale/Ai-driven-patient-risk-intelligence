# Step 1: Data Exploration (EDA)

As a **Healthcare Data Analyst**, the first step is always to understand your dataset. In this notebook, we will:
1. Load the synthetic patient data.
2. Check for missing values and data types.
3. Perform a simple join to connect Patients to their Diagnoses.

### Import Libraries

```python
import pandas as pd
import os

# Define path to data folder
DATA_DIR = '../data'

# Load datasets
df_patients = pd.read_csv(os.path.join(DATA_DIR, 'patients.csv'))
df_encounters = pd.read_csv(os.path.join(DATA_DIR, 'encounters.csv'))
df_diagnoses = pd.read_csv(os.path.join(DATA_DIR, 'diagnoses.csv'))
df_procedures = pd.read_csv(os.path.join(DATA_DIR, 'procedures.csv'))

print("Data loaded successfully!")
```

### 2. Analyze Diagnoses (ICD Codes)
Which diagnoses are most common in our patient population? This is a standard "Population Health" question.

```python
# Group by ICD code and count descriptions
top_diagnoses = df_diagnoses['icd_description'].value_counts().reset_index()
top_diagnoses.columns = ['Diagnosis', 'Count']
top_diagnoses
```

### 3. Join Tables (The "Analyst" Skill)
We need to see which patients have which conditions. In SQL, this would be a `LEFT JOIN`. In Pandas, it's `merge`.
Let's find all patients with Diabetes.

```python
# Filter for Diabetes (E11.9)
diabetes_codes = df_diagnoses[df_diagnoses['icd_code'].str.contains('E11')]

# Join with Patient Info
diabetic_patients = pd.merge(diabetes_codes, df_patients, on='patient_id', how='left')

print(f"Found {len(diabetic_patients)} diabetic encounters.")
diabetic_patients[['patient_id', 'first_name', 'last_name', 'icd_code', 'icd_description']].head()
```
